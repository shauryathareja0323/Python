```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Q1: What is Matplotlib? Why is it used? Name five plots that can be plotted using the Pyplot module of
#     Matplotlib.
# A1. Matplotlib is a comprehensive library in Python used for creating static, animated, and interactive visualizations. It is one of the most popular plotting libraries in the Python ecosystem and provides an easy way to generate plots, charts, and other types of visual representations of data.
#     Matplotlib is used for:
#     Data Visualization: It helps in visualizing data in a meaningful way, making it easier to understand patterns, trends, and outliers.
#     Customizable Plots: Users can customize almost every aspect of a plot, including colors, labels, axes, and legends.
#     Compatibility: Matplotlib integrates well with other libraries like NumPy, Pandas, and SciPy, making it a versatile tool in the data analysis workflow.
#     Wide Range of Plots: It offers a wide variety of plot types, allowing users to choose the best way to present their data.
#     Publication Quality: The plots created using Matplotlib can be customized to be of publication quality.
#
#     Five Plots That Can Be Plotted Using the Pyplot Module of Matplotlib
#     Line Plot (plot): A simple plot that shows the relationship between two variables by connecting data points with straight lines.
#     Bar Plot (bar): A chart that presents categorical data with rectangular bars, where the length of each bar is proportional to the value it represents.
#     Histogram (hist): A plot that shows the distribution of data by grouping the data into bins and displaying the frequency of each bin.
#     Scatter Plot (scatter): A plot that uses dots to represent the values obtained for two different variables. The position of each dot on the horizontal and vertical axis indicates values for an individual data point.
#     Pie Chart (pie): A circular plot divided into sectors, illustrating numerical proportions in a dataset.
```

```python
# Q2: What is a scatter plot? Use the following code to generate data for x and y. Using this generated data
#     plot a scatter plot.
# A2. Scatter plot consists of dots which are made on the intersection of data.
np.random.seed(3)
x = 3 + np.random.normal(0, 2, 50)
y = 3 + np.random.normal(0, 2, len(x))
plt.scatter(x,y)
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.title("Title")
```

```python
# Q3: Why is the subplot() function used? Draw four line plots using the subplot() function.
# A3. The subplot() function in Matplotlib is used to create multiple plots in a single figure. It allows you to arrange several plots in a grid, each with its own axes, within one figure window. This is particularly useful when you want to compare multiple plots side by side or visualize different aspects of the data simultaneously.
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 100, 200, 300, 400, 500])
plt.subplot(1,4,1)
plt.plot(x,y)

x = np.array([0, 1, 2, 3, 4, 5]) 
y = np.array([50, 20, 40, 20, 60, 70])
plt.subplot(1,4,2)
plt.plot(x,y)

x = np.array([0, 1, 2, 3, 4, 5]) 
y = np.array([10, 20, 30, 40, 50, 60])
plt.subplot(1,4,3)
plt.plot(x,y)

x = np.array([0, 1, 2, 3, 4, 5]) 
y = np.array([200, 350, 250, 550, 450, 150])
plt.subplot(1,4,4)
plt.plot(x,y)
```

```python
# Q4: What is a bar plot? Why is it used? Using the following data plot a bar plot and a horizontal bar plot.
# A4. A bar plot, also known as a bar chart, is a graphical representation of data where individual bars represent different categories or groups. The length or height of each bar is proportional to the value it represents. Bar plots can be displayed either vertically (vertical bar plot) or horizontally (horizontal bar plot).
company = np.array(["Apple", "Microsoft", "Google", "AMD"])
profit = np.array([3000, 8000, 1000, 10000])

plt.bar(company,profit)
```

```python
plt.barh(company,profit)
```

```python
# Q5: What is a box plot? Why is it used? Using the following data plot a box plot.
# A5. A box plot (or box-and-whisker plot) is a graphical representation of the distribution of a dataset. It shows the dataset's minimum, first quartile (Q1), median (Q2), third quartile (Q3), and maximum, along with potential outliers.
box1 = np.random.normal(100, 10, 200)
box2 = np.random.normal(90, 20, 200)

plt.boxplot(box1)
```

```python
plt.boxplot(box2)
```
