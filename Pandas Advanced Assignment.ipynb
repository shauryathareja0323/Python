```python
# Q4. Given a Pandas DataFrame df with a column 'Text', write a Python function to create a new column
#     'Word_Count' that contains the number of words in each row of the 'Text' column.
df=pd.DataFrame({"Text":["Hello Python","Hello Python is great","Python is a good programming language",""]})

def wordCount(df):
    l=[]
    length=len(df)
    for i in range(0,length):
        ele=df["Text"][i]
        l.append(len(ele.split()))
    
    df["Word_count"]=l
    return df
    

df=wordCount(df)
df
```

```python
# Q7. You have a Pandas DataFrame df that contains a column named 'Email' that contains email
#     addresses in the format 'username@domain.com'. Write a Python function that creates a new column
#     'Username' in df that contains only the username part of each email address.
import re
df=pd.DataFrame({"Email":["shaurya@g.com","shrey@mail.com","manu@outlook.com"]})

def username(df):
    r_num=df.shape[0]
    l=[]
    for i in range(0,r_num):
        email=df["Email"][i]
        pattern = r"^([^@]+)@"
        match = re.search(pattern, email)
        uname = match.group(1)
        l.append(uname)
    
    df["Username"]=l
    return df
        
        

df=username(df)
df
```

```python
# Q8. You have a Pandas DataFrame df with columns 'A', 'B', and 'C'. Write a Python function that selects
#     all rows where the value in column 'A' is greater than 5 and the value in column 'B' is less than 10. The
#     function should return a new DataFrame that contains only the selected rows.
df=pd.DataFrame({"A":[3,8,6,2,9],"B":[5,2,9,3,1],"C":[1,7,4,5,2]})

def func(df):
    n_rows=df.shape[0]
    l=[]
    for i in range(0,n_rows):
        a=df["A"][i]
        b=df["B"][i]
        if(a>5 and b<10):
            l.append(i)
    
    
    length=len(l)
    l1=[]
    l2=[]
    l3=[]
    for i in range(0,length):
        l1.append(df["A"][l[i]])
        l2.append(df["B"][l[i]])
        l3.append(df["C"][l[i]])
        
    df1=pd.DataFrame({"A":l1,"B":l2,"C":l3})
    return df1
        

print(df,"\n")
df1=func(df)
df1
```

```python
# Q9. Given a Pandas DataFrame df with a column 'Values', write a Python function to calculate the mean,
#     median, and standard deviation of the values in the 'Values' column.
df=pd.DataFrame({"Values":[10,20,30,40,50]})

def func(df):
    mean=df.mean()
    median=df.median()
    sd=df.std()
    
    print("Mean of the data is:",mean[0])
    print("Median of the data is:",median[0])
    print("Standard Deviation of the data is:",sd[0])

func(df)
```

```python
# Q11. You have a Pandas DataFrame df with a column 'Date'. Write a Python function that creates a new
#      column 'Weekday' in the DataFrame. The 'Weekday' column should contain the weekday name (e.g.
#      Monday, Tuesday) corresponding to each date in the 'Date' column.
import datetime

a = datetime.datetime(2024, 8, 24)
b = datetime.datetime(2024, 8, 25)
c = datetime.datetime(2024, 8, 26)
d = datetime.datetime(2024, 8, 27)
e = datetime.datetime(2024, 8, 28)

df=pd.DataFrame({"Date":[a,b,c,d,e]})

def weekday(df):
    n_rows=df.shape[0]
    l=[]
    for i in range(0,n_rows):
        x=df["Date"][i]
        l.append(x.strftime("%A"))
    
    df["Weekday"]=l
    return df
        

df=weekday(df)
df
```

```python
# Q12. Given a Pandas DataFrame df with a column 'Date' that contains timestamps, write a Python
#      function to select all rows where the date is between '2023-01-01' and '2023-01-31'.
import datetime
x=a.strftime("%A")
a = datetime.datetime(2023, 1, 24)
b = datetime.datetime(2023, 1, 25)
c = datetime.datetime(2023, 3, 26)
d = datetime.datetime(2023, 2, 27)
e = datetime.datetime(2023, 1, 28)

df=pd.DataFrame({"Date":[a,b,c,d,e]})

def date(df):
    s_date=datetime.datetime(2023, 1, 1)
    e_date=datetime.datetime(2023, 1, 31)
    n_rows=df.shape[0]
    l=[]
    for i in range(0,n_rows):
        x=df["Date"][i]
        if(x>s_date and x<e_date):
            l.append(i)
            
    return l
            

print("The rows where the Date lies between the given constraints are:",date(df))
df
```
