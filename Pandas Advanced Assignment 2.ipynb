```python
# Q2. What is the difference between the functions loc and iloc in pandas.DataFrame?
# A2. The loc() function is label based data selecting method which means that we have to pass the name
#     of the row or column which we want to select. This method includes the last element of the range passed 
#     in it, unlike iloc(). loc() can accept the boolean data unlike iloc(). The iloc() function is an
#     indexed-based selecting method which means that we have to pass an integer index in the method to select
#     a specific row/column. This method does not include the last element of the range passed in it unlike
#     loc(). iloc() does not accept the boolean data unlike loc(). 
```

```python
# Q9. Write a Python program that reads a CSV file containing categorical data and converts a specified
#     column to a categorical data type. The program should prompt the user to enter the file path, column
#     name, and category order, and then display the sorted data.

def read_and_convert_to_categorical():
   
    file_path = input("Enter the file path:")
    df = pd.read_csv(file_path)
    
    
   
    print("Columns in the CSV file:")
    print(df.columns)

    
    column_name = input("Enter the column name to convert to categorical: ")
    
    if column_name not in df.columns:
        print("Column not found. Please enter a valid column name.")
        return
    
    
    category_order = input("Enter the category order, separated by commas: ").split(',')
    category_order = [cat.strip() for cat in category_order]  # Clean up any spaces
    
   
    df[column_name] = pd.Categorical(df[column_name], categories=category_order, ordered=True)
    
   
    df_sorted = df.sort_values(by=column_name)
    
    
    print("\nSorted DataFrame:")
    print(df_sorted)


read_and_convert_to_categorical()
```

```python
# Q11. You are given a CSV file containing student data that includes the student ID and their test score. Write
#      a Python program that reads the CSV file, calculates the mean, median, and mode of the test scores, and
#      displays the results in a table.

def stats():
    filePath=input("Enter the file path")
    df=pd.read_csv(filePath)
    
    mean=df.mean()
    median=df.median()
    mode=df["Test Score"].mode()
    
    l1=[]
    l1.append(mean[1])
    l1.append(median[1])
    l1.append(mode[0])
    
    
    df1=pd.DataFrame({"Statistic":["mean","median","mode"],"Value":l1})
    return df1
    
df=stats()
df
```
